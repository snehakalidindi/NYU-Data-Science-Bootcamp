# 🧠 NumPy Basics: Step 1 – Array Creation & Properties
Welcome to your NumPy interview prep! Let's practice creating and inspecting arrays.

## ✅ Exercise 1: Create a 1D array from 10 to 19

```python
import numpy as np

# Your code here
arr1 = np.arange(10, 20)
print(arr1)
```

## ✅ Exercise 2: Create a 3x3 array filled with 7s

```python
# Your code here
arr2 = np.full((3, 3), 7)
print(arr2)
```

## ✅ Exercise 3: Generate 4 evenly spaced values from 0 to 1

```python
# Your code here
arr3 = np.linspace(0, 1, 4)
print(arr3)
```

## ✅ Exercise 4: Create a 2x5 array of random integers from 1 to 10

```python
# Your code here
arr4 = np.random.randint(1, 11, size=(2, 5))
print(arr4)
```

## ✅ Exercise 5: Explore array properties

```python
# Use one of the arrays above
print("Shape:", arr4.shape)
print("Size:", arr4.size)
print("Dimensions:", arr4.ndim)
print("Data Type:", arr4.dtype)
```
