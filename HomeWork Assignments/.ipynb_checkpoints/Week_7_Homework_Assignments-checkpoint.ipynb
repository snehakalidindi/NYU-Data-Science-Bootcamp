1. How do you assess the statistical significance of an insight?

I assess statistical significance using hypothesis testing—typically by computing a p-value. If the p-value is less than a chosen significance level (commonly 0.05), I reject the null hypothesis, indicating the result is statistically significant and not due to random chance.

```python
2. What is the Central Limit Theorem (CLT)? Why is it important?

The Central Limit Theorem states that the sampling distribution of the sample mean approaches a normal distribution as the sample size grows, regardless of the original population distribution—provided the sample size is large enough.

	•	Enables the use of normal-based methods (e.g., z-scores, confidence intervals) even when data isn’t normally distributed.
	•	Forms the foundation of many statistical inference techniques.
```

```python
3. What is statistical power?

Statistical power is the probability of correctly rejecting the null hypothesis when it is false (i.e., avoiding a Type II error).

A higher power means:
	•	Greater confidence in detecting a real effect.
	•	Usually achieved by increasing sample size or effect size.
```

```python
4. How do you control for biases?

To control for biases:
	•	Randomization: Assign subjects randomly to groups.
	•	Blinding: Prevent subjects or experimenters from knowing conditions.
	•	Stratification: Break groups into layers (e.g., by age/gender).
	•	Removing outliers or using robust statistics.
	•	Use matched-pair designs or covariate adjustments (e.g., regression).
```

```python
5. What are confounding variables?

A confounding variable is an external factor that is related to both the independent and dependent variables, potentially distorting the observed relationship.

Example: In studying coffee and heart disease, stress may be a confounder (those who drink more coffee might also be more stressed).
```

```python
6. What is A/B testing?

A/B testing is a controlled experiment where two versions (A and B) are compared:
	•	Group A sees the current version.
	•	Group B sees a variant.

We measure performance (like conversion rates) and use statistical tests to determine if the difference is significant.
```

7. What are confidence intervals?

A confidence interval gives a range of values likely to contain a population parameter (e.g., a mean) with a certain level of confidence—usually 95%.

Example: “We are 95% confident that the true mean salary lies between $65K and $72K.”
